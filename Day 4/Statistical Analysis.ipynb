```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])

print("Marks:", marks)
print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard Deviation:", np.std(marks))
print("Variance:", np.var(marks))
```

```python
import numpy as np

salaries = np.array([25000, 30000, 35000, 40000, 50000])

print("Average salary:", np.mean(salaries))
print("Highest salary:", np.max(salaries))
print("Lowest salary:", np.min(salaries))

above_average = salaries[salaries > np.mean(salaries)]
print("Salaries above average:", above_average)
```
