```python
import numpy as np
import time

numbers = list(range(100000))
arr = np.array(numbers)

# Using Python loop
start = time.time()

result = []
for n in numbers:
    result.append(n * 2)

loop_time = time.time() - start

# Using NumPy
start = time.time()

result = arr * 2

numpy_time = time.time() - start

print("Loop time:", loop_time)
print("NumPy time:", numpy_time)

# Performance Comparison: Loops vs Vectorization

# Objective:Compare Python loops with NumPy vectorization.

# Method:Multiplied 100,000 numbers by 2 using both methods and measured execution time.

# Observation:NumPy vectorization is often faster than Python loops for large numerical datasets.

# Conclusion:NumPy helps improve the speed and simplicity of numerical data processing.
```
