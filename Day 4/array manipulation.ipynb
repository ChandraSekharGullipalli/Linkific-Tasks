```python
# NumPy array
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print(arr)
```

```python
# array of zeros
arr = np.zeros(5)
print(arr)
```

```python
# array of ones
arr = np.ones((2, 3))
print(arr)
```

```python
# sequence of numbers
arr = np.arange(1, 11)
print(arr)
```

```python
# evenly spaced values
arr = np.linspace(0, 10, 5)
print(arr)
```

```python
# array shape
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)
```

```python
# number of dimensions
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.ndim)
```

```python
# one-dimensional array
arr = np.arange(1, 13)
reshaped = arr.reshape(3, 4)
print(reshaped)
```

```python
# Flatten a multidimensional array
arr = np.array([[1, 2], [3, 4]])
print(arr.flatten())
```

```python
# Transpose an array
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.T)
```

```python
# Access an array element
arr = np.array([10, 20, 30, 40, 50])
print(arr[2])
```

```python
# Slice an array
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr[1:5])
```

```python
# Change an array element
arr = np.array([10, 20, 30, 40])
arr[1] = 99
print(arr)
```

```python
# Append an element
arr = np.array([1, 2, 3])
new_arr = np.append(arr, 4)
print(new_arr)
```

```python
# Insert an element
arr = np.array([10, 20, 40, 50])
new_arr = np.insert(arr, 2, 30)
print(new_arr)
```

```python
# Delete an element
arr = np.array([10, 20, 30, 40])
new_arr = np.delete(arr, 1)
print(new_arr)
```

```python
# Concatenate two arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.concatenate((a, b))
print(result)
```

```python
# Stack arrays vertically
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(np.vstack((a, b)))
```

```python
# Stack arrays horizontally
a = np.array([[1], [2]])
b = np.array([[3], [4]])

print(np.hstack((a, b)))
```

```python
# Split an array
arr = np.array([10, 20, 30, 40, 50, 60])

parts = np.split(arr, 3)
print(parts)
```

```python
# Sort an array
arr = np.array([50, 10, 40, 20, 30])
print(np.sort(arr))
```

```python
# Find unique values
arr = np.array([1, 2, 2, 3, 3, 3, 4])
print(np.unique(arr))
```

```python
# Filter values using a condition
arr = np.array([10, 25, 30, 45, 50, 65])
filtered = arr[arr > 30]

print(filtered)
```

```python
# Replace values using a condition
arr = np.array([10, 20, 30, 40, 50])

arr[arr < 30] = 0
print(arr)
```

```python
# Copy an array safely
original = np.array([10, 20, 30])
copied = original.copy()

copied[0] = 100

print("Original:", original)
print("Copied:", copied)
```
