```python
import pandas as pd

data = {
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Employee": [
        "Ravi", "Priya", "Arun", "Sneha", "Kiran",
        "Meena", "Rahul", "Divya", "Ajay", "Pooja"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "Finance", "IT", "HR", "Finance", "IT"
    ],
    "City": [
        "Hyderabad", "Bengaluru", "Chennai", "Hyderabad",
        "Pune", "Chennai", "Bengaluru", "Hyderabad",
        "Pune", "Chennai"
    ],
    "Salary": [
        45000, 35000, 55000, 60000, 40000,
        50000, 65000, 38000, 52000, 48000
    ],
    "Experience": [2, 1, 4, 5, 2, 3, 6, 2, 4, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
#Group by Department
df.groupby("Department").groups
```

```python
# Group employees by Department
df.groupby("Department")["Employee"].apply(list)
```

```python
# Group by Department and calculate average salary
df.groupby("Department")["Salary"].mean()
```

```python
# Group by City
df.groupby("City")["Employee"].apply(list)
```

```python
# Group by multiple columns
df.groupby(["Department", "City"])["Salary"].mean()
```

```python
# Count employees in each department
df.groupby("Department").size()
```

```python
# Group by Department and show all employee details
for department, group in df.groupby("Department"):
    print("\nDepartment:", department)
    print(group)
```

```python
# Aggregation examples
# Calculate total salary
df["Salary"].sum()
```

```python
# Calculate average salary
df["Salary"].mean()
```

```python
# Find the highest salary
df["Salary"].max()
```

```python
# Find the lowest salary
df["Salary"].min()
```

```python
# Count employees
df["Employee"].count()
```

```python
# Calculate total experience
df["Experience"].sum()
```

```python
# Perform multiple aggregations
df["Salary"].agg(["sum", "mean", "min", "max", "count"])
```

```python
# Aggregate multiple columns    
df.agg({
    "Salary": ["sum", "mean", "max"],
    "Experience": ["sum", "mean", "min"]
})
```
