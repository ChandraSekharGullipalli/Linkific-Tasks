```python
# create a DataFrame
import pandas as pd

data = {
    "EmployeeID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [45000, 35000, 55000, 60000, 40000],
    "City": ["Hyderabad", "Bengaluru", "Chennai", "Hyderabad", "Pune"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Explore the dataset
print(df.head())
print(df.tail())
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.info())
print(df.describe())
```

```python
# Select rows and columns
# Select one column
print(df["Name"])

# Select multiple columns
print(df[["Name", "Salary"]])

# Select rows by position
print(df.iloc[0:3])

# Select rows by condition
print(df.loc[df["Department"] == "IT"])
```

```python
# Filter data
# Employees earning more than 40000
print(df[df["Salary"] > 40000])

# Employees from Hyderabad
print(df[df["City"] == "Hyderabad"])

# Employees in IT with salary above 40000
print(df[(df["Department"] == "IT") & (df["Salary"] > 40000)])

# Employees in HR or Finance
print(df[df["Department"].isin(["HR", "Finance"])])
```

```python
# Add, update, and delete columns
# Add a new column
df["Bonus"] = df["Salary"] * 0.10

# Update a column
df["Salary"] = df["Salary"] + 2000

# Create a derived column
df["AnnualSalary"] = df["Salary"] * 12

print(df)
```

```python
# Handle missing values
df_missing = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sneha"],
    "Salary": [45000, None, 55000, None],
    "Department": ["IT", "HR", None, "Finance"]
})

print(df_missing.isnull().sum())

# Fill missing salaries with the median
df_missing["Salary"] = df_missing["Salary"].fillna(
    df_missing["Salary"].median()
)

# Fill missing departments
df_missing["Department"] = df_missing["Department"].fillna("Unknown")

print(df_missing)
```

```python
# Remove duplicate records
duplicate_df = pd.DataFrame({
    "EmployeeID": [101, 102, 102, 103],
    "Name": ["Ravi", "Priya", "Priya", "Arun"],
    "Salary": [45000, 35000, 35000, 55000]
})

print("Duplicates:", duplicate_df.duplicated().sum())

clean_df = duplicate_df.drop_duplicates()

print(clean_df)
```

```python
# Sort and rank data
# Sort by salary in descending order
print(df.sort_values(by="Salary", ascending=False))

# Rank employees by salary
df["SalaryRank"] = df["Salary"].rank(
    ascending=False,
    method="dense"
)

print(df[["Name", "Salary", "SalaryRank"]])
```

```python
# Group and aggregate data
# Average salary by department
print(df.groupby("Department")["Salary"].mean())

# Multiple calculations
summary = df.groupby("Department").agg(
    EmployeeCount=("EmployeeID", "count"),
    AverageSalary=("Salary", "mean"),
    MaximumSalary=("Salary", "max"),
    MinimumSalary=("Salary", "min")
)

print(summary)
```

```python
# Apply functions and transform text
# Convert names to uppercase
df["NameUpper"] = df["Name"].str.upper()

# Clean extra spaces in text
df["City"] = df["City"].str.strip()

# Apply a function to salaries
df["SalaryCategory"] = df["Salary"].apply(
    lambda salary: "High" if salary >= 50000 else "Regular"
)

print(df)
```

```python
# Merge two DataFrames
employees = pd.DataFrame({
    "EmployeeID": [101, 102, 103],
    "Name": ["Ravi", "Priya", "Arun"]
})

departments = pd.DataFrame({
    "EmployeeID": [101, 102, 103],
    "Department": ["IT", "HR", "Finance"]
})

merged_df = pd.merge(
    employees,
    departments,
    on="EmployeeID",
    how="inner"
)

print(merged_df)
```

```python
# Concatenate DataFrames
team1 = pd.DataFrame({
    "Name": ["Ravi", "Priya"],
    "Department": ["IT", "HR"]
})

team2 = pd.DataFrame({
    "Name": ["Arun", "Sneha"],
    "Department": ["Finance", "IT"]
})

combined_df = pd.concat(
    [team1, team2],
    ignore_index=True
)

print(combined_df)
```

```python
# Work with dates
df_dates = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "JoiningDate": ["2024-01-15", "2023-06-20", "2025-03-10"]
})

df_dates["JoiningDate"] = pd.to_datetime(df_dates["JoiningDate"])

df_dates["JoiningYear"] = df_dates["JoiningDate"].dt.year
df_dates["JoiningMonth"] = df_dates["JoiningDate"].dt.month

print(df_dates)
```

```python
# Read and write CSV files
# Read the CSV file
employees_df = pd.read_csv("employees.csv")

print(employees_df.head())

# Filter the data
high_salary_df = employees_df[
    employees_df["Salary"] > 40000
]

# Save the filtered data
high_salary_df.to_csv(
    "high_salary_employees.csv",
    index=False
)

print("CSV file created successfully!")
```

```python
# Create a final data analysis report
report = employees_df.groupby("Department").agg(
    TotalEmployees=("EmployeeID", "count"),
    TotalSalary=("Salary", "sum"),
    AverageSalary=("Salary", "mean")
).reset_index()

print(report)

report.to_csv("department_report.csv", index=False)
```
